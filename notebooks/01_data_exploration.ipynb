```python
#dropping the columns and updating the dataframe
cols_to_drop = ['Product Description','Order Zipcode', 'Product Image','Customer Password','Customer Email']
df = df.drop(columns=cols_to_drop)
```

```python
df.shape
```

```python
#remove
#1. whitespace
#2. lowercase
#3. remove currency symbol
#4. replace space with underscore

df.columns = (df.columns
                    .str.strip()
                    .str.lower()
                    .str.replace(r'[()]', '', regex=True)
                    .str.replace(r'\s+', '_', regex=True)
)
```

```python
#rename columns
df = df.rename(columns={'order_date_dateorders': 'order_date', 'shipping_date_dateorders': 'shipping_date'})
```

```python
df.columns.tolist()
```

```python
#convert order_date and shipping_date to datetime format
df['order_date'] = pd.to_datetime(df['order_date'])
df['shipping_date'] = pd.to_datetime(df['shipping_date'])
```

```python
#calculating shipping duration in days
days_for_shipping_real = (df['shipping_date'] - df['order_date']).dt.days
days_for_shipment_scheduled = df['days_for_shipment_scheduled']

df['shipping_delay_days'] = days_for_shipping_real - days_for_shipment_scheduled
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month

for days in df['shipping_delay_days']:
    if days < 0:
        print("Early Delivery")
    elif days == 0:
        print("On Time Delivery")
    else:
        print("Late Delivery")
```

```python
#viewing the first 5 rows of the relevant columns
df[['order_date', 'shipping_date', 'days_for_shipping_real',
    'days_for_shipment_scheduled', 'shipping_delay_days',
    'late_delivery_risk']].head(5)
```

```python
#calculated column
import numpy as np

#create conditions and choices for the new column
conditions = [
    df['shipping_delay_days'] < 0,
    df['shipping_delay_days'] == 0,
    df['shipping_delay_days'] > 0
]

#create choices for the new column
choices = [
    'Early Delivery',
    'On Time Delivery',
    'Late Delivery'
]

#create a new column 'delivery_performance' based on the conditions and choices
df['delivery_performance'] = np.select(conditions, choices, default='Unknown')
```

```python
#an executive summary table showing:

#Each delivery_performance category ('Late Delivery', 'Early Delivery', 'On Time Delivery').
#The order_count (number of orders in each category).
#The percentage of total orders that each category represents, rounded to 2 decimal places.
#Sorted in descending order by order_count.
query = """
SELECT delivery_performance,
       COUNT(*) AS order_count,
       ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM orders), 2) AS percentage
FROM orders
GROUP BY delivery_performance
ORDER BY order_count DESC
"""
pd.read_sql_query(query, conn)
```

```python
#We want to build a table grouped by shipping_mode showing:

#shipping_mode
#total_orders: Total count of orders.
#avg_scheduled_days: Average scheduled delivery days (rounded to 2 decimals).
#avg_actual_days: Average real delivery days (rounded to 2 decimals).
#avg_delay_days: Average delay duration (rounded to 2 decimals).
#late_rate_pct: Percentage of orders that were late (rounded to 2 decimals).
#Sorted from highest late rate to lowest (late_rate_pct DESC).

query = """
SELECT shipping_mode,
       COUNT(*) AS total_orders,
       ROUND(AVG(days_for_shipment_scheduled), 2) AS avg_scheduled_days,
       ROUND(AVG(days_for_shipping_real), 2) AS avg_actual_days,
       ROUND(AVG(shipping_delay_days), 2) AS avg_delay_days,
       ROUND(SUM(CASE WHEN shipping_delay_days > 0 THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS late_rate_pct
FROM orders
GROUP BY shipping_mode
ORDER BY late_rate_pct DESC
"""
pd.read_sql_query(query, conn)
```

```python
#Write a query to display:

#order_region
#market
#total_orders: Count of orders
#total_sales: ROUND(SUM(sales), 2)
#avg_delay_days: ROUND(AVG(shipping_delay_days), 2)
#late_risk_pct: ROUND(AVG(late_delivery_risk) * 100.0, 2)
#Filter to keep only regions with COUNT(*) > 5000 using HAVING.
#Order by late_risk_pct DESC.

query = """
SELECT order_region,
       market,
       COUNT(*) AS total_orders,
       ROUND(SUM(sales), 2) AS total_sales,
       ROUND(AVG(shipping_delay_days), 2) AS avg_delay_days,
       ROUND(AVG(late_delivery_risk) * 100.0, 2) AS late_risk_pct
FROM orders
GROUP BY order_region, market
HAVING COUNT(*) > 5000
ORDER BY late_risk_pct DESC
"""
pd.read_sql_query(query, conn)
```

```python
#Top 3 worst delayed product categories in each market
query = """
WITH category_delays AS (
    SELECT market,
           category_name,
           COUNT(*) AS total_orders,
           ROUND(AVG(shipping_delay_days), 2) AS avg_delay_days,
           ROUND(AVG(late_delivery_risk) * 100.0, 2) AS late_risk_pct,
           ROW_NUMBER() OVER (
               PARTITION BY market 
               ORDER BY AVG(late_delivery_risk) DESC
           ) AS rank_in_market
    FROM orders
    GROUP BY market, category_name
    HAVING COUNT(*) >= 100
)
SELECT market,
       category_name,
       total_orders,
       avg_delay_days,
       late_risk_pct,
       rank_in_market
FROM category_delays
WHERE rank_in_market <= 3;
"""

pd.read_sql_query(query, conn)
```
